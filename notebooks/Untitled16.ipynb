```python
import json
import subprocess
import pandas as pd

# Initialize metadata collection
metadata_results = []

# Download and inspect metadata for all participants
for participant_id in participants["participant_id"]:

    s3_path = (
        f"s3://openneuro.org/ds005892/{participant_id}/func/"
        f"{participant_id}_task-rest_bold.json"
    )

    local_path = "/content/temp_bold.json"

    result = subprocess.run(
        [
            "aws", "s3", "cp", s3_path,
            local_path, "--no-sign-request"
        ],
        capture_output=True,
        text=True
    )

    if result.returncode == 0:
        with open(local_path, "r") as file:
            metadata = json.load(file)

        metadata_results.append({
            "participant_id": participant_id,
            "TR": metadata.get("RepetitionTime"),
            "TE": metadata.get("EchoTime"),
            "TaskName": metadata.get("TaskName"),
            "Manufacturer": metadata.get("Manufacturer"),
            "MagneticFieldStrength": metadata.get(
                "MagneticFieldStrength"
            )
        })

    else:
        metadata_results.append({
            "participant_id": participant_id,
            "TR": None,
            "TE": None,
            "TaskName": None,
            "Manufacturer": None,
            "MagneticFieldStrength": None
        })

# Create metadata table
metadata_df = pd.DataFrame(metadata_results)

print("FUNCTIONAL MRI METADATA")
display(metadata_df)

# Check unique acquisition parameters
print("\nUNIQUE TECHNICAL VALUES")

for column in [
    "TR",
    "TE",
    "TaskName",
    "Manufacturer",
    "MagneticFieldStrength"
]:
    print(f"\n{column}:")
    print(metadata_df[column].value_counts(dropna=False))

print("\nTotal participants:", len(metadata_df))
print("Successfully inspected:",
      metadata_df["TR"].notna().sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import shift

# Load BOLD image
bold_img = nib.load("/content/sub-MJF001_bold.nii.gz")
bold_data = bold_img.get_fdata()

# Calculate frame-to-frame difference
n_volumes = bold_data.shape[3]

frame_difference = []

for i in range(1, n_volumes):
    diff = np.mean(
        np.abs(bold_data[:, :, :, i] - bold_data[:, :, :, i-1])
    )
    frame_difference.append(diff)

# Plot results
plt.figure(figsize=(12, 5))
plt.plot(frame_difference)
plt.title("Frame-to-Frame BOLD Signal Difference")
plt.xlabel("Volume")
plt.ylabel("Mean Absolute Difference")
plt.grid()
plt.show()

# Summary
print("Number of volumes:", n_volumes)
print("Mean frame difference:", np.mean(frame_difference))
print("Maximum frame difference:", np.max(frame_difference))
print("Standard deviation:", np.std(frame_difference))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from nilearn.signal import clean

# Original BOLD data
bold_img = nib.load("/content/sub-MJF001_bold.nii.gz")
data = bold_img.get_fdata()

# Reshape: voxels x time -> time x voxels
time_data = data.reshape(-1, data.shape[3]).T

# Temporal filtering
filtered_data = clean(
    time_data,
    detrend=True,
    standardize="zscore_sample",
    low_pass=0.1,
    high_pass=0.01,
    t_r=2.0
)

# Restore original dimensions
filtered_data = filtered_data.T.reshape(data.shape)

# Save results
np.save(
    "/content/filtered_bold.npy",
    filtered_data
)

# Compare original and filtered signals
voxel_signal = data[32, 32, 16, :]
filtered_signal = filtered_data[32, 32, 16, :]

plt.figure(figsize=(12, 5))

plt.plot(voxel_signal, label="Original signal")
plt.plot(filtered_signal, label="Filtered signal")

plt.title("Original vs Filtered BOLD Signal")
plt.xlabel("Time points")
plt.ylabel("Signal intensity")
plt.legend()
plt.grid()
plt.show()

print("Original shape:", data.shape)
print("Filtered shape:", filtered_data.shape)
print("Saved: filtered_bold.npy")
```

```python

import nibabel as nib
import numpy as np
from nilearn import datasets, image

# Load mean BOLD
mean_bold = nib.load(
    "/content/sub-MJF001_mean_bold.nii.gz"
)

# Load Harvard-Oxford atlas
atlas = datasets.fetch_atlas_harvard_oxford(
    "cort-maxprob-thr25-2mm"
)
atlas_img = atlas.maps

# Resample atlas to BOLD voxel grid
atlas_resampled = image.resample_to_img(
    atlas_img,
    mean_bold,
    interpolation="nearest"
)

# Save resampled atlas
atlas_resampled.to_filename(
    "/content/atlas_resampled_to_bold.nii.gz"
)

# Check results
print("Mean BOLD shape:", mean_bold.shape)
print("Resampled atlas shape:", atlas_resampled.shape)

print("\nSame affine:",
      np.allclose(mean_bold.affine,
                  atlas_resampled.affine))

atlas_data = atlas_resampled.get_fdata()

print("\nUnique atlas labels in BOLD space:",
      np.unique(atlas_data))

print("\nNumber of non-background voxels:",
      np.sum(atlas_data > 0))
```

```python

import nibabel as nib
import numpy as np
import pandas as pd
from nilearn.maskers import NiftiLabelsMasker

# Load original BOLD image
bold_img = nib.load("/content/sub-MJF001_bold.nii.gz")

# Load resampled Harvard-Oxford atlas
atlas_img = nib.load(
    "/content/atlas_resampled_to_bold.nii.gz"
)

# Create ROI masker
masker = NiftiLabelsMasker(
    labels_img=atlas_img,
    standardize="zscore_sample",
    detrend=True,
    t_r=2.0,
    verbose=1
)

# Extract ROI time series
time_series = masker.fit_transform(bold_img)

# Get atlas labels (excluding background)
roi_labels = atlas.labels[1:]

# Create DataFrame
time_df = pd.DataFrame(
    time_series,
    columns=roi_labels
)

# Save results
time_df.to_csv(
    "/content/roi_time_series.csv",
    index=False
)

print("Time series shape:", time_series.shape)
print("Number of time points:", time_series.shape[0])
print("Number of ROIs:", time_series.shape[1])

print("\nFirst 5 rows:")
print(time_df.iloc[:5, :5])
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load ROI time series
time_df = pd.read_csv("/content/roi_time_series.csv")

# Calculate Pearson correlation matrix
correlation_matrix = time_df.corr(method="pearson")

# Save correlation matrix
correlation_matrix.to_csv(
    "/content/functional_connectivity_matrix.csv"
)

# Display matrix information
print("Correlation matrix shape:",
      correlation_matrix.shape)

print("\nFirst 5x5 values:")
print(correlation_matrix.iloc[:5, :5].round(3))

# Visualize the matrix
plt.figure(figsize=(14, 12))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    interpolation="nearest"
)

plt.colorbar(label="Pearson correlation")
plt.title("Functional Connectivity Matrix")
plt.xlabel("Brain Regions (ROI)")
plt.ylabel("Brain Regions (ROI)")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90,
    fontsize=7
)

plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index,
    fontsize=7
)

plt.tight_layout()
plt.savefig(
    "/content/functional_connectivity_matrix.png",
    dpi=300
)
plt.show()

print("\nMatrix saved successfully!")
```
